```python
import pandas as pd
import numpy as np
```

```python
train = pd.read_csv('C:\Users\twoth\Downloads\train.csv')
train
train_x = train.iloc[:,0:7]
train_x
train_y = train.iloc[:,7]
train_y
test_x = pd.read_csv('C:\Users\twoth\Downloads/test.csv')
sample_submission = pd.read_csv('C:\Users\twoth\Downloadssample_submission.csv',index_col = 0)
sample_submission
#train_x, train_y = train.iloc[:, :-1], train.iloc[:, -1] 이렇게 함으로써 간결하게 표현가능하다.
```

```python
#1등 코드 
from sklearn.ensemble import RandomForestClassifier
model = RandomForestClassifier(
    n_estimators=100, 
    max_depth=8, 
)
model.fit(train_x,train_y)
y = model.predict(test_x)
y
```

```python
from sklearn.neighbors import KNeighborsClassifier
kn = KNeighborsClassifier(weights='distance',leaf_size=20,algorithm='kd_tree')
kn.fit(train_x, train_y.values.ravel())
kn.score(train_x, train_y.values.ravel())
y = kn.predict(test_x)
y
```

```python
sample_submission['target'] =y
sample_submission.to_csv('sample_submission.csv')
sample_submission
```
